```python
                                         
# cheack wheather a triangle is right-angled

a= float(input("enter side 1: "))
b= float(input("enter side 2: "))
c= float(input("enter side 3: "))

print("sides 1:", a)
print("sides 2:", b)
print("sides 3:", c) 

if a**2 + b**2 == c**2:
    print("the tringle is right-angled.")
else:
    print("the tringle is not right-angled.")
    

```
